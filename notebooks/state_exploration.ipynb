```python
import pandas as pd

df = pd.read_csv("../data/processed/master_dataset.csv")

print(df.shape)

df.head()
```

```python
df["speed_delta"] = df["speed"].diff()

df["speed_delta"].describe()
```

```python
df["rpm_delta"] = df["rpm"].diff()

df["rpm_delta"].describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
df["speed_delta"].hist(bins=100)
plt.title("Speed Delta Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df["rpm_delta"].hist(bins=100)
plt.title("RPM Delta Distribution")
plt.show()
```
